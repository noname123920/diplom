# Сетка 100 узлов:

```octave
clear; close all;
L = 1.0; kappa = 1.0;
Nx = 100;
dx = L / (Nx - 1);
x = linspace(0, L, Nx);
dt = 0.4 * (dx^2 / kappa);
t_target = 0.01;
Nt = round(t_target / dt);
p = zeros(1, Nx);
p(1) = 1;
p(end) = 0;

for n = 1:Nt
    p_inner = p(2:end-1) + (kappa * dt / dx^2) * (p(3:end) - 2*p(2:end-1) + p(1:end-2));
    p = [1, p_inner, 0];
end
p_100 = p;
```

# Сетка в два раз мельче:

```octave
L = 1.0; kappa = 1.0;
Nx = 200;
dx = L / (Nx - 1);
x = linspace(0, L, Nx);
dt = 0.4 * (dx^2 / kappa);
t_target = 0.01;
Nt = round(t_target / dt);
p = zeros(1, Nx);
p(1) = 1;
p(end) = 0;

for n = 1:Nt
    p_inner = p(2:end-1) + (kappa * dt / dx^2) * (p(3:end) - 2*p(2:end-1) + p(1:end-2));
    p = [1, p_inner, 0];
end
p_200 = p;
```

```octave
size(p_100)
size(p_200)
```

```octave
p_100_interp = interp1(linspace(0, L, 100), p_100, linspace(0, L, 200));
```

```octave
size(p_100_interp)
```

```octave
abs_err = abs(p_200 - p_100_interp);
```

```octave
p_200(:);
```

```octave
rel_err = abs_err ./ max(max(p_200, p_100_interp), 1e-10) * 100; % массив из погрешностей для каждой точки в процентах
size(rel_err)
```

```octave
max_rel_err = max(rel_err) % условию удовлетворяет
```
